```python
# find the fasta for BRCA1 homo sapiens fasta from ncbi

from Bio import SeqIO

record = SeqIO.read("sequence.fasta", "fasta")

print("ID: ", record.id)
print("Description: ", record.description)
print("Sequence: ", record.seq)
print("Length: ", len(record.seq))
```

```python
# find the sequence length

print("Sequence Length: ", len(record.seq))
print("Number of (A): ", record.count("A"))
print("Number of (T): ", record.count("T"))
print("Number of (C): ", record.count("C"))
print("Number of (G): ", record.count("G"))

gc_count = record.count("G") + record.count("C")
gc_content = (gc_count / len(record.seq)) * 100
print("GC_Content: ", gc_content, "%")
```
