```python
sequence = "ATGCGATCGATCGAT"

print(sequence)
```

```python
# write python code to find GC_content

print("Length: ", len(sequence))
print("A: ", sequence.count("A"))
print("T: ", sequence.count("T"))
print("C: ", sequence.count("C"))
print("G: ", sequence.count("G"))

gc_count = sequence.count("G") + sequence.count("C")
gc_content = (gc_count / len(sequence)) * 100

print("GC_Content: ", gc_content, "%")
```

```python
from Bio.Seq import Seq

dna = Seq("ATGCGATCGATCGAT")

print(dna)
```

```python
print(type(sequence))
print(type(dna))
```

```python
# first letter of the sequence

print("First: ", dna[0])

# last letter of the sequence

print("Last: ", dna[-1])
```

```python
# Transription

rna = dna.transcribe()

print("DNA: ", dna)
print("RNA: ", rna)
```

```python
# Translation

protein = rna.translate()

print("Protein: ", protein)
```

```python
# reverse transcription - converts an RNA sequence back to dna

print("Reverse transcription DNA: ", rna.back_transcribe())
```
