```python
# represent an alignment containing multiple sequence
from Bio.Align import MultipleSeqAlignment

# represent a biological sequence
from Bio.Seq import Seq

# store a sequence toghter with its identifier and its related
from Bio.SeqRecord import SeqRecord

```

```python
records = [
    SeqRecord(Seq("ATGCT"), id = "Seq1"),
    SeqRecord(Seq("ATGTT"), id = "Seq2"),
    SeqRecord(Seq("ATGCC"), id = "Seq3")
]

for i in records:
    print(i.id, ":", i.seq)
```

```python
# create an display an alignment object
alignment = MultipleSeqAlignment(records)
print(alignment)
```

```python
# Display the sequnces position by position
print("Position: ", " ".join(str(i) for i in range(1, 6)))

for record in alignment:
    print(record.id + ":", " ".join(record.seq))
```

```python
# calculate conserved and variable positions
conserved_positions = []
variable_positions = []

for i in range(alignment.get_alignment_length()):
    # extracts the nucleotide at position i
    column = [str(record.seq[i]) for record in alignment]
    
    # collects the distinct nucleotides in that column
    if len(set(column)) == 1:
        conserved_positions.append(i + 1)
    else:
        variable_positions.append(i + 1)
        
print("Conserved positions: ", conserved_positions)
print("Variable positions: ", variable_positions)
```

```python
# calculate the percentage of consevered positions

total_positions = alignment.get_alignment_length()

#(3/5)*100
conserved_percentage = (
    len(conserved_positions) / total_positions
) * 100

print(f"Conserved positions: {conserved_percentage: .2f}%")
```

```python
# generate the consensus sequence
from collections import Counter

consensus = ""

for i in range(alignment.get_alignment_length()):
    column = [str(record.seq[i]) for record in alignment]
    counts = Counter(column)
    
    most_common_base, count = counts.most_common(1)[0]
    consensus += most_common_base
    
print("Consensus sequence: ", consensus)
```

```python
# inspect the nucleotide frequncies
# calculate percentage identity

a = consensus.count("A")
t = consensus.count("T")
g = consensus.count("G")
c = consensus.count("C")

total_nucleotides = len(consensus)
print("A frequency: ", a, "/", total_nucleotides)
```
