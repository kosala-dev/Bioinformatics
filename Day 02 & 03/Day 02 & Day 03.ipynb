```python
base = input("Enter a nucleotide(A/T/C/G): ")

if base == "A":
    print("Adenine")
elif base == "T":
    print("Thymine")
elif base == "C":
    print("Cytosine")
elif base == "G":
    print("Guanine")
else:
    print("Invalid nucleotide")
```

```python
# DNA validation

sequence = "CTTTCTGTCXCGCCCTTCCTCTGA"

if "X" in sequence:
    print("Invalid DNA Sequence")
else:
    print("Valid DNA Sequence")
```

```python
# another way to validate

sequence = "CTTTCTGTCCGCCCTTCCTCTGA"

valid_bases = "ATCG"

valid = True

for base in sequence:
    if base not in valid_bases:
        valid = False
        
if valid:
    print("Valid DNA Sequence")
else :
    print("Invalid DNA Sequence")
```

```python
# Dataframes

import pandas as pd

data = {
    "Gene": ["BRCA1", "TP53", "EGFR", "KRAS"],
    "Organism": ["Human", "Human", "Human", "Human"],
    "length": [1200, 1500, 1100, 980],
    "GC_Content":[45.2, 51.3,48.7,42.1]
}

df = pd.DataFrame(data)

print(df)
```
